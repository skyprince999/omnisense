```python
import videodb
from videodb import IndexType

conn = videodb.connect(VIDEODB_API_KEY)
coll = conn.get_collection()

scene_index_id = []
scene_index_name = []

for video in coll.get_videos():
    print(f"\nVideo {video.id} — {video.name}")
    for idx in video.list_scene_index():
        print(f"  scene_index_id={idx['scene_index_id']}  "
              f"status={idx.get('status')}  name={idx.get('name', 'unnamed')}")
        scene_index_id.append(idx['scene_index_id'])
        scene_index_name.append(idx.get('name', 'unnamed'))
```

```python
results = video.search(
    "harsh braking",
    index_type=IndexType.scene,
    index_id=scene_index_id[0],   # from list_scene_index() above
)
for shot in results.get_shots():
    print(shot.start, shot.end, shot.text)
```
